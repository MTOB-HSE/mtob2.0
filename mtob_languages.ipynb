```python
import pandas as pd

df = pd.read_csv(r"/content/drive/MyDrive/from_mteb_to_mtob - грамматики.csv", header=None)
```

```python
# Define your new column names here
new_column_names = [
    "Title",
    "Language_Family",
    "Language",
    "Continent",
    "Author",
    "Year",
    "Pages"
] # Example names, please adjust as needed

# Assign new column names to the DataFrame
df.columns = new_column_names

# Display the DataFrame with new headers
df.head()
```

```python
# add ISO codes

#from glottolog https://github.com/glottolog/glottolog

!git clone https://github.com/glottolog/glottolog
```

```python
glotto = pd.read_csv(r"/content/drive/MyDrive/languoid.csv")

glotto.head()
```

```python
len(glotto)
```

```python
glottolog = {}
for i in range(len(glotto)):
  glottolog[glotto.iloc[i]["name"].lower()] = glotto.iloc[i]["iso639P3code"]
```

```python
isos = []
for i in range(len(df)):
  if df.iloc[i]['Language'].lower() in glottolog:
    isos.append(glottolog[df.iloc[i]['Language'].lower()])
  else:
    isos.append('no exact match')
```

```python
df['iso639P3code'] = isos
```

```python
df.to_csv(
    "/content/drive/MyDrive/updated_mtob.csv"
)
```

## SMOL

221 low resource languages -- MT to English

https://huggingface.co/datasets/google/smol

```python
smol = pd.read_csv(r"/content/drive/MyDrive/smol_langs.csv")

smol.head()
```

```python
smol_iso = {}
for i in range(len(smol)):
  smol_iso[smol.iloc[i]["ISO639-3"]] = smol.iloc[i]["LP"]
```

```python
isos = []
for i in range(len(df)):
  if df.iloc[i]['iso639P3code'] in smol_iso:
    isos.append(smol_iso[df.iloc[i]['iso639P3code']])
  else:
    isos.append('')
```

```python
df['smol'] = isos
```

```python
df.to_csv(
    "/content/drive/MyDrive/updated_mtob.csv"
)
```
